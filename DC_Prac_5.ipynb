```python
# =====================================================
# MAPREDUCE PROGRAM
# FIND HOTTEST AND COOLEST YEAR
# =====================================================

# Sample Weather Data
# Format:
# Date  MinTemp  MaxTemp

data = """2020-01-01 10 25
2020-05-10 20 35
2021-02-15 5 20
2021-06-20 25 40
2022-03-10 15 30
2022-07-15 28 45"""


# =====================================================
# MAP PHASE
# =====================================================

mapped = []

for line in data.split("\n"):

    parts = line.strip().split()

    # Skip invalid lines
    if len(parts) != 3:
        continue

    date, min_temp, max_temp = parts

    # Extract year from date
    year = date.split("-")[0]

    # Store mapped values
    mapped.append((year, int(min_temp), int(max_temp)))


# =====================================================
# REDUCE PHASE
# =====================================================

from collections import defaultdict

temps_by_year = defaultdict(list)

# Group temperatures by year
for year, min_t, max_t in mapped:

    temps_by_year[year].append((min_t, max_t))


# =====================================================
# PROCESS RESULTS
# =====================================================

result = {}

for year in temps_by_year:

    mins = [t[0] for t in temps_by_year[year]]
    maxs = [t[1] for t in temps_by_year[year]]

    # Find minimum and maximum temperature
    result[year] = (min(mins), max(maxs))


# =====================================================
# FIND HOTTEST AND COOLEST YEAR
# =====================================================

coolest_year = min(result, key=lambda x: result[x][0])

hottest_year = max(result, key=lambda x: result[x][1])


# =====================================================
# OUTPUT
# =====================================================

print("===================================")
print("YEAR-WISE WEATHER DATA")
print("===================================\n")

for year in sorted(result):

    print(year,
          "Min Temp:", result[year][0],
          "Max Temp:", result[year][1])


print("\n===================================")
print("FINAL RESULT")
print("===================================\n")

print("Coolest Year :", coolest_year)

print("Hottest Year :", hottest_year)
```
