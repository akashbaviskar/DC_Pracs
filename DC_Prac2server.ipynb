```python
import Pyro4

@Pyro4.expose
class StringConcatenator(object):

    def concatenate(self, str1, str2):
        return str1 + " " + str2

# Create daemon
daemon = Pyro4.Daemon()

# Register object
uri = daemon.register(StringConcatenator)

print("Server URI :", uri)

print("Server is running...")

# Start server loop
daemon.requestLoop()
```
