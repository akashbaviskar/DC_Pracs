```python
from collections import defaultdict
with open("sample.txt", "w") as file:
    file.write("Hello World\n")
    file.write("MapReduce Example\n")
    file.write("Distributed Computing\n")
    file.write("Hello MapReduce")
    
def char_mapper(text):

    mapped = []

    for char in text:

        if char != " " and char != "\n":
            mapped.append((char, 1))

    return mapped

def word_mapper(lines):

    mapped = []

    for line in lines:

        words = line.strip().split()

        for word in words:
            mapped.append((word.lower(), 1))

    return mapped

def reducer(mapped_data):

    reduced = defaultdict(int)

    for key, value in mapped_data:
        reduced[key] += value

    return reduced

with open("sample.txt", "r") as file:

    text = file.read()

with open("sample.txt", "r") as file:

    lines = file.readlines()

char_mapped = char_mapper(text)

# REDUCE PHASE
char_result = reducer(char_mapped)

print("===================================")
print("CHARACTER COUNT")
print("===================================\n")

for char, count in sorted(char_result.items()):
    print(f"{char} : {count}")

word_mapped = word_mapper(lines)

word_result = reducer(word_mapped)

print("\n===================================")
print("WORD COUNT")
print("===================================\n")

for word, count in sorted(word_result.items()):
    print(f"{word} : {count}")
```
